```python
import json
def load_data(filename):
    with open(filename,'r') as file:
        data=json.load(file)
    return data
def pages_may_ulike(user_id,data):
    user_pages={}
    for user in data["users"]:
        user_pages[user["id"]]=set(user["liked_pages"])
    if user_id not in user_pages:
        return []
    user_liked_page=user_pages[user_id]
    page_suggestions={}
    for other_user,pages in user_pages.items():
        if other_user != user_id:
            sharing_pages=user_liked_page.intersection(pages)
        for page in pages:
            if page not in user_liked_page:
                page_suggestions[page]=page_suggestions.get(page,0)+len(sharing_pages)
    # print(page_suggestions)
    sorted_page=sorted(page_suggestions.items(), key=lambda x:x[1],reverse=True)
 
    return [(page,count) for page,count in sorted_page]
    
data=load_data("data_people_uk.json")
user_id=1
sugg=pages_may_ulike(user_id,data)
print(sugg)
```
