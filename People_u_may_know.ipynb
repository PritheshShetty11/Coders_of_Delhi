```python
import json
def load_data(filename):
    data=json.load(open(filename,'r'))
    return data
def People_u_may_know(user_id,data):
    user_friends={}
    for user in data["users"]:
        user_friends[user["id"]]=list(set(user["friends"]))
    if user_id not in user_friends:
        return []
    direct_frnds=user_friends[user_id]
    suggestions={}
    for friend in direct_frnds:
        for mutual in user_friends[friend]:
            if mutual!=user_id and mutual not in direct_frnds:
                suggestions[mutual]=suggestions.get(mutual,0)+1
    sorted_suggestions=sorted(suggestions.items(),key=lambda x:x[1],reverse=True)
    # print(sorted_suggestions)
    return [user for user,connections in sorted_suggestions ]
                
        


data=load_data('data_people_uk.json')
user_id=1
recc=People_u_may_know(user_id,data)
print(recc)
```
