```python
import json
def load_file(filename):
    with open(f"{filename}","r") as f:
        data=json.load(f)
    return data
data=load_file("data.json")
```

```python
def display_users(data):
    for user in data['users']:
        print(f"{user['id']} is {user['name']}")
    for page in data['pages']:
        print(f"{page['id']} is {page['name']}")

display_users(data)
```
